```python
from pathlib import Path
import pandas as pd
import numpy as np

# Locate the project directory
PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

# Dataset path
DATA_PATH = PROJECT_ROOT / "data" / "raw"
file_path = DATA_PATH / "DataCoSupplyChainDataset.csv"

# Load dataset
df = pd.read_csv(file_path, encoding="latin1")

print("Dataset loaded successfully!")
print("-" * 50)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("File size:", round(file_path.stat().st_size / (1024 * 1024), 2), "MB")
```

```python
# Display first 5 records
display(df.head())

# Display all column names
print("\nCOLUMN NAMES:")
print(df.columns.tolist())

# Check data types and non-null counts
print("\nDATASET INFORMATION:")
df.info()
```

```python
# 1. Missing values
missing = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean() * 100).round(2)
})

missing = missing.sort_values(
    by="Missing Percentage",
    ascending=False
)

display(missing[missing["Missing Count"] > 0])

# 2. Duplicate records
print("Exact duplicate rows:", df.duplicated().sum())

# 3. Data types
print("\nData types:")
display(df.dtypes.to_frame("Data Type"))

# 4. Numerical summary
print("\nNumerical summary:")
display(df.describe().T)
```

```python
# Check unique values in important categorical columns
categorical_columns = [
    "Delivery Status",
    "Order Status",
    "Shipping Mode",
    "Customer Segment",
    "Market"
]

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))

# Check numerical ranges
numeric_columns = [
    "Benefit per order",
    "Sales per customer",
    "Sales",
    "Order Profit Per Order",
    "Order Item Quantity",
    "Days for shipping (real)",
    "Days for shipment (scheduled)"
]

display(df[numeric_columns].describe().T)
```

```python
numeric_columns = [
    "Benefit per order",
    "Sales per customer",
    "Sales",
    "Order Profit Per Order",
    "Order Item Quantity",
    "Days for shipping (real)",
    "Days for shipment (scheduled)"
]

summary = df[numeric_columns].describe().T.round(2)

display(summary)
```

```python
print("Negative benefit records:",
      (df["Benefit per order"] < 0).sum())

print("Negative profit records:",
      (df["Order Profit Per Order"] < 0).sum())

print("Zero or negative quantities:",
      (df["Order Item Quantity"] <= 0).sum())

print("Negative sales records:",
      (df["Sales"] < 0).sum())

print("Maximum actual shipping days:",
      df["Days for shipping (real)"].max())

print("Maximum scheduled shipping days:",
      df["Days for shipment (scheduled)"].max())
```

```python
# Profitability by product category
profit_by_category = (
    df.groupby("Category Name")
    .agg(
        Total_Sales=("Sales", "sum"),
        Average_Profit=("Benefit per order", "mean"),
        Total_Profit=("Benefit per order", "sum"),
        Records=("Order Item Id", "count")
    )
    .sort_values("Total_Profit")
)

display(profit_by_category.head(10))
```

```python
# Delivery performance by shipping mode
shipping_analysis = (
    df.groupby("Shipping Mode")
    .agg(
        Average_Actual_Days=("Days for shipping (real)", "mean"),
        Average_Scheduled_Days=("Days for shipment (scheduled)", "mean"),
        Late_Records=("Late_delivery_risk", "sum"),
        Total_Records=("Order Item Id", "count")
    )
)

shipping_analysis["Late_Percentage"] = (
    shipping_analysis["Late_Records"] /
    shipping_analysis["Total_Records"] * 100
).round(2)

display(shipping_analysis.round(2))
```

```python
# Compare delivery status with late delivery risk
delivery_check = pd.crosstab(
    df["Delivery Status"],
    df["Late_delivery_risk"],
    margins=True
)

display(delivery_check)
```

```python
# Compare actual and scheduled shipping days
df["Shipping_Delay_Days"] = (
    df["Days for shipping (real)"]
    - df["Days for shipment (scheduled)"]
)

print("Shipping delay distribution:")
display(df["Shipping_Delay_Days"].value_counts().sort_index())

print("\nAverage delay by shipping mode:")
display(
    df.groupby("Shipping Mode")["Shipping_Delay_Days"]
    .agg(["mean", "median", "min", "max"])
    .round(2)
)
```

```python
# Delivery status against calculated shipping delay
delay_status_check = pd.crosstab(
    df["Shipping_Delay_Days"],
    df["Delivery Status"]
)

display(delay_status_check)
```

```python
# Check whether the two profit columns are identical
profit_comparison = (
    df["Benefit per order"] - df["Order Profit Per Order"]
)

print("Maximum absolute difference:",
      profit_comparison.abs().max())

print("Are both columns identical?",
      df["Benefit per order"].equals(df["Order Profit Per Order"]))
```

```python
profit_comparison = (
    df["Benefit per order"] -
    df["Order Profit Per Order"]
)

print("Maximum absolute difference:",
      profit_comparison.abs().max())

print(
    "Are both columns identical?",
    df["Benefit per order"].equals(
        df["Order Profit Per Order"]
    )
)
```

```python
# Delivery status and shipping delay for canceled records
canceled = df[df["Delivery Status"] == "Shipping canceled"]

print("Canceled records:", len(canceled))

display(
    canceled["Shipping_Delay_Days"]
    .value_counts()
    .sort_index()
)
```

```python
# Convert date columns in a separate working copy

clean_df = df.copy()

date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

for col in date_columns:
    clean_df[col] = pd.to_datetime(
        clean_df[col],
        format="mixed",
        errors="coerce"
    )

# Check date conversion
for col in date_columns:
    print(f"\n{col}")
    print("Data type:", clean_df[col].dtype)
    print("Missing dates:", clean_df[col].isna().sum())
    print("Earliest date:", clean_df[col].min())
    print("Latest date:", clean_df[col].max())
```

```python
# Columns to remove from analytical dataset

columns_to_drop = [
    "Product Description",       # 100% missing
    "Order Profit Per Order",    # Exact duplicate of Benefit per order
    "Customer Email",
    "Customer Fname",
    "Customer Lname",
    "Customer Password",
    "Customer Street"
]

clean_df = clean_df.drop(columns=columns_to_drop)

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", clean_df.shape)

print("\nRemoved columns:")
print(columns_to_drop)

print("\nRemaining columns:", len(clean_df.columns))
```

```python
# Recreate working copy from original raw dataset
clean_df = df.copy()

# Convert date columns again
date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

for col in date_columns:
    clean_df[col] = pd.to_datetime(
        clean_df[col],
        format="mixed",
        errors="coerce"
    )

print("Working copy recreated:", clean_df.shape)
```

```python
columns_to_drop = [
    "Product Description",
    "Order Profit Per Order",
    "Customer Email",
    "Customer Fname",
    "Customer Lname",
    "Customer Password",
    "Customer Street"
]

# Identify columns that actually exist
existing_columns = [
    col for col in columns_to_drop
    if col in clean_df.columns
]

# Remove them
clean_df = clean_df.drop(columns=existing_columns)

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", clean_df.shape)

print("\nColumns removed:")
print(existing_columns)

print("\nRemaining columns:", len(clean_df.columns))
```

```python
# Missing values after initial cleaning

missing_summary = pd.DataFrame({
    "Missing Count": clean_df.isnull().sum(),
    "Missing Percentage": (
        clean_df.isnull().mean() * 100
    ).round(2)
})

missing_summary = missing_summary.sort_values(
    "Missing Percentage",
    ascending=False
)

display(
    missing_summary[
        missing_summary["Missing Count"] > 0
    ]
)
```

```python
# Remove highly incomplete Order Zipcode column
clean_df = clean_df.drop(columns=["Order Zipcode"])

print("Order Zipcode removed.")
print("Customer Zipcode missing:",
      clean_df["Customer Zipcode"].isnull().sum())

print("Current dataset shape:", clean_df.shape)
```

```python
# Check unique business identifiers

print("Unique Orders:", clean_df["Order Id"].nunique())

print("Unique Customers:", clean_df["Customer Id"].nunique())

print("Unique Order Items:", clean_df["Order Item Id"].nunique())

print("Duplicate Order IDs:", clean_df["Order Id"].duplicated().sum())

print("Duplicate Order Item IDs:", clean_df["Order Item Id"].duplicated().sum())
```

```python
# Extract useful date features

clean_df["Order_Year"] = (
    clean_df["order date (DateOrders)"].dt.year
)

clean_df["Order_Month"] = (
    clean_df["order date (DateOrders)"].dt.month
)

clean_df["Order_Month_Name"] = (
    clean_df["order date (DateOrders)"].dt.month_name()
)

clean_df["Order_Quarter"] = (
    clean_df["order date (DateOrders)"].dt.quarter
)

clean_df["Shipping_Delay_Days"] = (
    clean_df["Days for shipping (real)"] -
    clean_df["Days for shipment (scheduled)"]
)

# Check new features
display(
    clean_df[
        [
            "order date (DateOrders)",
            "Order_Year",
            "Order_Month",
            "Order_Month_Name",
            "Order_Quarter",
            "Shipping_Delay_Days"
        ]
    ].head()
)

print("Updated dataset shape:", clean_df.shape)
```

```python
from pathlib import Path

# Create processed data folder if needed
processed_path = PROJECT_ROOT / "data" / "processed"
processed_path.mkdir(parents=True, exist_ok=True)

# Save cleaned dataset
output_file = processed_path / "supplychain_cleaned.csv"

clean_df.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("Location:", output_file)
print("Shape:", clean_df.shape)
```
